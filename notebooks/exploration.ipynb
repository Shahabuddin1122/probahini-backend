```python
def _transliterate_banglish(query: str) -> str:
        """Transliterate Banglish (Bangla in Latin script) to Bangla script."""
        # Use indic_transliteration to convert Banglish to Bangla
        try:
            return transliterate(query, sanscript.ITRANS, sanscript.BENGALI)
        except Exception as e:
            # Fallback: return original query if transliteration fails
            print(f"Transliteration error: {e}")
            return query
```

```python
_transliterate_banglish(query="mashik er somoy ki kora jay")
```
